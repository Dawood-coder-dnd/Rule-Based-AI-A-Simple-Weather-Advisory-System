```python
def predict_next_day(past_temperatures, window=3):
    """Predicts tomorrow's temperature as the average of the last 'window' days."""
    recent=past_temperatures[-window:]
    prediction=sum(recent)/len(recent)
    return round(prediction,1)
if __name__ =="__main__":
    past_week = [33,40,17,38,-10,19]
    actual_tomorrow = 35 # what the temperature actually turned out to be 
    predicted = predict_next_day(past_week, window=3)
    error = round(abs(predicted - actual_tomorrow),1)
    print(f"Predicted: {predicted}C")
    print(f"Actual:  {actual_tomorrow}C")
    print(f"Error:  {error}C")
```

```python
 def weather_advice(temperature_c, humidity_percent, wind_speed_kmh, air_quality_index):
    """Rule-based weather advisor using simple IF/THEN logic."""
    advice = []

    # Temperature rules
    if temperature_c >= 35:
        advice.append("Extreme heat: stay hydrated and avoid the sun.")
    elif temperature_c <= 5:
        advice.append("Cold weather: Wear a warm jacket.")

    # Humidity rule
    if humidity_percent >= 80:
        advice.append("High humidity: Expect possible rain, carry an umbrella.")

    # New Rule 1: Wind speed
    if wind_speed_kmh >= 40:
        advice.append("Strong winds: Be careful outdoors and secure loose objects.")

    # New Rule 2: Air quality
    if air_quality_index >= 150:
        advice.append("Poor air quality: Avoid prolonged outdoor activities.")

    # Default advice
    if not advice:
        advice.append("Conditions look moderate. No special precautions needed.")

    return advice


if __name__ == "__main__":
    sample_days = [
        (20, 21, 10, 50),
        (6, 66, 15, 80),
        (27, 85, 20, 100),
        (38, 70, 45, 170),
        (2, 60, 50, 200),
        (25, 50, 10, 40)
    ]

    for temp, humidity, wind, aqi in sample_days:
        result = weather_advice(temp, humidity, wind, aqi)

        print("----------------------------------------")
        print(f"Temperature: {temp}°C")
        print(f"Humidity: {humidity}%")
        print(f"Wind Speed: {wind} km/h")
        print(f"Air Quality Index: {aqi}")
        print("Advice:")

        for message in result:
            print(f"- {message}")
```

```python
def spam_classifier(email_text):
    spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= 3:
        return "Spam"
    else:
        return "Legitimate"


if __name__ == "__main__":
    test_emails = [
        ("Congratulations! You are a winner! Get your free prize, click here now!", "Spam"),
        ("Urgent! You have won a free prize. Click here to claim your reward.", "Spam"),
        ("You are the lucky winner. Get your free prize today!", "Spam"),
        ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
        ("Your assignment is due on Friday. Please submit it on time.", "Legitimate"),
        ("Thank you for your purchase. Your order has been shipped.", "Legitimate")
    ]

    correct = 0

    for email, correct_label in test_emails:
        predicted_label = spam_classifier(email)

        print("----------------------------------------")
        print("Email:", email)
        print("Correct Label:", correct_label)
        print("Predicted Label:", predicted_label)

        if predicted_label == correct_label:
            print("Result: Correct")
            correct += 1
        else:
            print("Result: Incorrect")

    accuracy = (correct / len(test_emails)) * 100

    print("----------------------------------------")
    print(f"Correct Predictions: {correct}/{len(test_emails)}")
    print(f"Accuracy: {accuracy:.2f}%")

# Like facial recognition and medical image diagnosis, this classifier assigns an input to one of a fixed set of categories.
```

```python
def spam_classifier(email_text, threshold):
    spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

    email_text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in email_text:
            count += 1

    if count >= threshold:
        return "Spam"
    else:
        return "Legitimate"


if __name__ == "__main__":
    test_emails = [
        ("Congratulations! You are a winner! Get your free prize, click here now!", "Spam"),
        ("Urgent! You have won a free prize. Click here to claim your reward.", "Spam"),
        ("You are the lucky winner. Get your free prize today!", "Spam"),
        ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
        ("Your assignment is due on Friday. Please submit it on time.", "Legitimate"),
        ("Thank you for your purchase. Your order has been shipped.", "Legitimate")
    ]

    for threshold in range(1, 6):
        correct = 0

        for email, correct_label in test_emails:
            predicted_label = spam_classifier(email, threshold)

            if predicted_label == correct_label:
                correct += 1

        accuracy = (correct / len(test_emails)) * 100

        print(f"Threshold = {threshold}")
        print(f"Correct = {correct}/{len(test_emails)}")
        print(f"Accuracy = {accuracy:.2f}%")
        print("----------------------------------------")
```
